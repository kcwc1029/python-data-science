# NumPy：從陣列到高效數值運算


## 認識 NumPy：一週餐費一起算

下班買便當、早餐店加蛋、下午買飲料，一筆金額不難算；如果要把整週、整班或整家店的數字一起調整，反覆寫加減乘除就容易漏掉。NumPy 適合讓「同一條計算規則」作用在一整批數值上。

`import numpy as np` 是載入套件並取簡稱；`np` 是慣例，不是 Python 關鍵字。`np.array(...)` 把現有數值建立成陣列。`餐費 * 2` 的意思要看餐費是哪一種物件：List 重複內容，數值陣列則讓每個數字乘二。這叫「逐元素運算」，元素就是其中一格的值。

| 工具      | 適合情境                         | 需要留意                           |
| --------- | -------------------------------- | ---------------------------------- |
| List      | 購物項目、少量混合資料           | 乘法可能是重複，不是金額翻倍       |
| ndarray   | 同一批價格、成績、數量做相同運算 | 主要依位置配對，自己維護位置的意義 |
| DataFrame | 有姓名、日期、類別的表格         | 運算可能按標籤對齊                 |


````python
"""認識 NumPy：一週餐費一起算。對應 numpy.md 第 01 節。"""

import numpy as np
import pandas as pd

# 五天午餐費；List 的乘二會重複整份清單。
lunch_list = [90, 110, 100, 120, 80]
print("List 乘二：", lunch_list * 2)

# ndarray 的乘二是逐格計算金額。
lunch = np.array(lunch_list)
print("陣列乘二：", lunch * 2)
print("每天少花十元：", lunch - 10)
print("整週午餐費：", lunch.sum())

# 同一批數字也可以加上星期標籤，交給 Pandas 呈現。
report = pd.DataFrame({
    "星期": ["一", "二", "三", "四", "五"], 
    "餐費": lunch
    })
print(report)
````

NumPy 常能把大量數值運算交給底層實作，減少 Python 逐次執行迴圈的負擔。但小資料、轉換成本、複雜文字處理都可能讓它沒有速度優勢；先追求計算正確與可讀性，不要把「用了 NumPy」當作一定比較快。

## ndarray：一維、二維與多維陣列

「維度」可以先理解成：要用幾個位置，才能找到一個數字。一維是排成一列的記錄；二維要指定第幾列與第幾欄；三維再增加第幾張表。

以飲料銷量為例，一天三種飲料是 `(3,)`；兩天、每天三種飲料是 `(2, 3)`；兩間店、每間兩天、每天三種飲料是 `(2, 2, 3)`。三維不需要想成抽象立體空間，只要想成兩份相同格式的表格。

我們固定把「列」當橫向一排（row），把「欄」當直向一欄（column）。

````python
"""ndarray：一維、二維與多維陣列。對應 numpy.md 第 02 節。"""

import numpy as np

# 一維：紅茶、奶茶、綠茶，順序固定。
one_day = np.array([20, 15, 10])

# 二維：列是日期，欄是飲料。
one_shop = np.array([[20, 15, 10], [30, 10, 20]])

# 三維：店別、日期、飲料。
two_shops = np.array([one_shop, [[10, 5, 8], [15, 8, 12]]])

print("一天：", one_day.shape)
print("一店兩天：", one_shop.shape)
print("兩店：", two_shops.shape)
# 索引從 0 開始：第二間、第一天、第三種飲料。
print("指定銷量：", two_shops[1, 0, 2])
````

## 資料型態與轉換：數字外觀不等於數字

````python
"""看起來像數字，不代表資料型態是數字"""
import numpy as np

prices = np.array(["120", "85", "99.5"])

print(prices) # ['120' '85' '99.5']
print(prices.dtype) # <U4，表示文字型態
````

````python
"""文字無法直接當成數字運算"""
prices = np.array(["120", "85", "99.5"])
# prices + 10 # 會發生錯誤
````

必須先將文字轉換成數值型態

- 使用 astype(float) 轉換成小數
- astype() 預設會建立轉換後的陣列

````python
prices = np.array(["120", "85", "99.5"])
numeric_prices = prices.astype(float)

print(numeric_prices) # [120.   85.   99.5]
print(numeric_prices.dtype) # float64
````

````python
import numpy as np

# 字串中的純數字可轉為浮點數；「待確認」則不能直接 astype。
price_text = np.array(["120", "85.5", "99.9"])
prices = price_text.astype(float)
print("九折：", prices * 0.9)
print("截去小數：", np.array([3.8, -3.8]).astype(int))
# 先轉浮點數，才能保留修改後的小數。
editable = np.array([100, 200]).astype(float)
editable[0] = 89.5
print("修改後：", editable)
print("浮點比較：", np.isclose(0.1 + 0.2, 0.3))
print("取最近偶數：", np.round([2.5, 3.5]))
````

## 布林遮罩

布林遮罩(Boolean Mask)可以把它想成一張「篩選表」。

### 建立布林遮罩

如果想知道「哪些商品價格超過 100 元」，可以直接寫：

````python
import numpy as np

### 建立商品價格
prices = np.array([50, 120, 80, 200, 150])

### 判斷每個價格是否大於 100
mask = prices > 100

print(mask)
````

### 使用遮罩篩選資料

````python
import numpy as np


### 建立商品價格
prices = np.array([50, 120, 80, 200, 150])
mask = prices > 100 ### 建立布林遮罩

### 使用遮罩篩選
result = prices[mask]
print(result)
````

### 多個條件同時篩選

| 運算子  | 意思 | 範例                                 |
| ---- | -- | ---------------------------------- |
| `&`  | 而且 | `(prices >= 80) & (prices <= 150)` |
| `\|` | 或者 | `(prices < 80) \| (prices > 150)`  |
| `~`  | 反向 | `~(prices > 100)`                  |

例如現在想找「價格介於 80 元到 150 元之間的商品」，就寫`(prices >= 100) & (prices <= 180)`。

注意，不是`(prices >= 80) and (prices <= 150)`，因為 Python 的 and、or 處理的是「整體 True / False」。NumPy 需要的是逐元素判斷


````python
import numpy as np

### 建立商品價格
prices = np.array([50, 120, 80, 200, 150])

### 價格大於等於 80，而且小於等於 150
mask = (prices >= 80) & (prices <= 150)

print(prices[mask])
````

### mask.sum()：有幾筆符合條件
布林值在加總時：
```
True  → 1
False → 0
```
因此可以直接計算符合條件的商品數量

````python
import numpy as np

### 建立商品價格
prices = np.array([50, 120, 80, 200, 150])

### 找出價格超過 100 的商品
mask = prices > 100

print(mask)

### 計算符合條件的數量
count = mask.sum()
print("超過 100 元的商品數量：", count)
````

### any()：有沒有任何一筆符合

如果只想知道「有沒有商品超過 180 元？」

````python
import numpy as np


### 建立商品價格
prices = np.array([50, 120, 80, 200, 150])


### 判斷是否至少有一個商品超過 180 元
mask = prices > 180

print(mask)
print(mask.any())
````

### all()：是不是全部都符合
如果只想知道「所有商品是不是都至少 50 元？」

````python
import numpy as np


### 建立商品價格
prices = np.array([50, 120, 80, 200, 150])


### 判斷所有商品是否都大於等於 50
mask = prices >= 50

print(mask)
print(mask.all())
````

### 商品名稱與價格一起篩選
布林遮罩真正好用的地方，是可以同步篩選不同陣列。

````python
import numpy as np


### 建立商品資料
products = np.array([
    "紅茶",
    "珍珠奶茶",
    "雞排",
    "牛肉麵",
    "咖啡"
])

prices = np.array([
    30,
    70,
    90,
    180,
    120
])


### 找出價格大於等於 100 的商品
mask = prices >= 100


### 使用相同遮罩篩選商品名稱與價格
selected_products = products[mask]
selected_prices = prices[mask]

print(selected_products)
print(selected_prices)
````

### 直接使用遮罩修改原始資料

遮罩不只能「找資料」，還可以直接修改資料。

例如「200 元以上的商品全部打九折」



````python
import numpy as np


### 建立商品價格
prices = np.array([80, 120, 200, 250, 300], dtype=float)


### 找出 200 元以上的商品
mask = prices >= 200


### 將符合條件的商品打九折
prices[mask] = prices[mask] * 0.9

print(prices)
````

## 修改陣列：view 與 copy

### 最基本的情況：b = a

可以想成：
```
a ──┐
    ├──→ [100, 120, 150, 180] -> 它們都指向同一個 NumPy 陣列物件
b ──┘
```

### 用 is 檢查是不是同一個物件

Python 可以用 is 判斷兩個變數是不是指向同一個物件：

````python
import numpy as np

a = np.array([100, 120, 150, 180])
b = a

### 判斷是否為同一個物件
print(a is b)
````

### 切片比較特別：會產生 view


````python
import numpy as np


### 建立原始價格
a = np.array([100, 120, 150, 180])


### 取出前兩筆資料
b = a[:2]

print("a：", a)
print("b：", b)
print(a is b) # False

# 看起來 a、b 已經不是同一個物件。這時候的b是 view(檢視)。
````

### View 是什麼

```
原始資料 a
┌───────────────────────┐
│ 100 │ 120 │ 150 │ 180 │
└───────────────────────┘
```

```
b = a[:2]

原始資料 a
┌───────────────────────┐
│ 100 │ 120 │ 150 │ 180 │
└─▲─────▲───────────────┘
  │     │
  └── b ┘
```
所以 b 雖然是一個新的陣列物件，底下的資料仍然來自 a。

````python
# 修改 view，原始資料也會改
import numpy as np


### 建立原始價格
a = np.array([100, 120, 150, 180])

b = a[:2] ### 取得前兩筆資料的 view
b[0] = 50 ### 修改 b

print("a：", a)
print("b：", b)
````

### copy()：真正複製一份資料

如果你的目的就是「拿資料出來試算，但不要動到原始資料」，就使用`.copy()`

````python
import numpy as np

a = np.array([100, 120, 150, 180]) ### 建立原始價格
b = a[:2].copy() ### 複製前兩筆資料
b[0] = 50 ### 修改複製後的資料

print("a：", a)
print("b：", b)
````

## 調整形狀：重新排版，不是重新排序

````python
# 假設記錄了連續 14 天的餐費：
import numpy as np


### 建立 14 天餐費
expenses = np.array([
    120, 150, 100, 180, 130, 200, 160,
    140, 170, 110, 190, 150, 220, 180
])

print(expenses)
print("形狀：", expenses.shape) # 目前是一維陣列，總共有 14 個元素。
````

### reshape()：改變陣列形狀


````python
import numpy as np


### 建立 14 天餐費
expenses = np.array([
    120, 150, 100, 180, 130, 200, 160,
    140, 170, 110, 190, 150, 220, 180
])

### 重新排列成 2 週，每週 7 天
weekly = expenses.reshape(2, 7)

print(weekly)
print("形狀：", weekly.shape)
````

## 陣列合併與拆分

「你是要把資料「接長」，還是要多增加一個「分類層級」？」，這就是 concatenate() 和 stack() 最大的差別。

````python
import numpy as np


### 建立早班銷售資料
morning = np.array([
    [10, 20, 15],
    [12, 18, 22]
])

print(morning)
print("形狀：", morning.shape)
````

### concatenate()：沿著原本的方向接起來


````python
import numpy as np


### 前兩天
first = np.array([
    [10, 20, 15],
    [12, 18, 22]
])


### 後兩天
second = np.array([
    [14, 25, 17],
    [16, 21, 20]
])

result = np.concatenate(
    [first, second],
    axis=0 # 依「列(row)」方向增加資料
    # axis=1 # 依「欄(column)」方向增加資料
)

print(result)
print("形狀：", result.shape)
````

### stack()：不是接長，而是新增一個軸

````python
# 早班
morning = np.array([ # 2 天 × 3 種商品
    [10, 20, 15],
    [12, 18, 22]
])

# 中班
evening = np.array([ # 2 天 × 3 種商品
    [15, 25, 20],
    [18, 22, 30]
])
````

````python
# 如果直接用concatenate()合併成(4, 3)資料，班別這個結構消失了。
sales = np.concatenate(
    [morning, evening],
    axis=0
)

print(sales)
print("\n形狀：")
print(sales.shape)
````

````python
# 用 stack() 保留「班別」這個層級
### 新增班別這個軸
sales = np.stack(
    [morning, evening],
    axis=0
)

print(sales)

print("\n形狀：")
print(sales.shape)
````

## 向量化：把共同規則寫一次

假設團購三種餐點：

````python
import numpy as np

prices = np.array([90, 110, 80]) ### 商品單價
quantities = np.array([2, 1, 3]) ### 購買數量
````

現在要算每項商品的小計

````python
# 傳統寫法可以用迴圈
total = 0
for price, quantity in zip(prices, quantities):
    total += price * quantity
print(total)
````

````python
### NumPy 可以直接逐元素計算商品小計
subtotals = prices * quantities

print(subtotals)
````

## Broadcasting：廣播機制：不同形狀如何配對
假設三戶家庭分別購買牛奶與雞蛋：

````python
import numpy as np

### 三戶家庭的購買數量
quantities = np.array([
    [2, 1],
    [1, 3],
    [4, 2]
])

### 牛奶與雞蛋單價
prices = np.array([80, 60]) # 這是一微陣列，如果要以二維振烈去看，就是(1, 2)

print("數量形狀：", quantities.shape)
print("價格形狀：", prices.shape)
````

````python
### 計算每戶各商品小計
subtotals = quantities * prices
print(subtotals)
````

Broadcasting會從從 shape 的右邊開始比較。
```
(3, 2)
(1, 2)
右邊都是2，所以可以計算
```

只要滿足以下兩個條件即可：
- 兩邊數字相同 → 可以
- 其中一邊是 1 → 可以

如果想要再增加折扣

````python
discounts = np.array([
    0.9,
    1.0,
    0.8
])

print(discounts.shape) # (3,)
# print(subtotals * discounts) # (3,2) 與 (1, 3)無法Broadcasting
````

````python
### 處理方式，把把折扣變成 (3, 1)
discounts = np.array([
    0.9,
    1.0,
    0.8
])

discounts = discounts.reshape(3, -1)
print(discounts.shape)
print(subtotals * discounts) 
````

## 數學函式與條件運算

### np.where()：像 Excel 的 IF

````python
# 假設三筆訂單：
### 訂單金額
orders = np.array([320, 500, 780])
### 根據訂單金額決定運費
shipping = np.where(orders >= 500, 0, 60)
print(shipping)
````

### np.clip()：限制上下限

````python
# 假設會員點數
points = np.array([ -5, 30, 150])

# 規定：
# 最低 0
# 最高 100

### 將點數限制在 0～100
result = np.clip(
    points,
    0,
    100
)

print(result)
````

### np.abs()：只看差多少

````python
# 假設目標消費是 500：
orders = np.array([
    320,
    500,
    780
])

### 計算與 500 的距離
difference = np.abs(orders - 500)
print(difference)
````

### np.sqrt()：平方根

````python
import numpy as np

### 計算平方根
values = np.array([ 4, 9, 16])
result = np.sqrt(values)
print(result)
````
