### Problem 建立自己的 Series

某間飲料店記錄三種飲料今天的銷售杯數：

| 飲料 | 杯數 |
| -- | -: |
| 紅茶 | 50 |
| 綠茶 | 35 |
| 奶茶 | 60 |


請使用 pd.Series() 建立一個名為 sales 的 Series：

- values 為 50、35、60
- index 為 "紅茶"、"綠茶"、"奶茶"

最後印出 sales。

```python
import pandas as pd

sales = pd.Series(
    [50, 35, 60],
    index=["紅茶", "綠茶", "奶茶"]
)

print(sales)
```
