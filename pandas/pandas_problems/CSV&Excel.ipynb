### Problem. 處理 CSV 後重新輸出

- 讀取 '../pandas_datasets/orders.csv'
- 新增「總金額」欄位。
- 總金額 = 數量 × 單價。
- 將處理後的資料輸出成 clean.csv。
- 不要輸出 DataFrame 的 index。
- 使用 utf-8-sig 編碼輸出。

```python
import pandas as pd

df = pd.read_csv(
    "../pandas_datasets/orders.csv",
    encoding="utf-8"
)

df["總金額"] = df["數量"] * df["單價"]

df.to_csv(
    "clean.csv",
    index=False,
    encoding="utf-8-sig"
)
```

### Problem

- 讀取 "../pandas_datasets/sales.csv"。
- 新增「總金額」欄位，計算 數量 × 單價。
- 將總金額 大於等於 3000 的資料存入 high_sales。
- 將總金額 小於 3000 的資料存入 normal_sales。
- 將全部處理完成的資料輸出成 clean_sales.csv，使用 utf-8-sig，且不要輸出 index。
- 將 high_sales 與 normal_sales 輸出到同一個 sales_report.xlsx。
- Excel 中建立「高額訂單」與「一般訂單」兩個工作表。
- Excel 中不要輸出 DataFrame 的 index。

```python
import pandas as pd

# 讀取 CSV
df = pd.read_csv(
    "../pandas_datasets/sales.csv",
    encoding="utf-8"
)

# 計算總金額
df["總金額"] = df["數量"] * df["單價"]

# 分類
high_sales = df[df["總金額"] >= 3000]
normal_sales = df[df["總金額"] < 3000]

# 輸出 CSV
df.to_csv(
    "clean_sales.csv",
    index=False,
    encoding="utf-8-sig"
)

# 輸出 Excel
with pd.ExcelWriter(
    "sales_report.xlsx",
    engine="openpyxl"
) as writer:

    high_sales.to_excel(
        writer,
        sheet_name="高額訂單",
        index=False
    )

    normal_sales.to_excel(
        writer,
        sheet_name="一般訂單",
        index=False
    )
```
