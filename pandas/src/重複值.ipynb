# 重複值
## 建立範例資料

```python
import pandas as pd


### 建立交易資料
data = {
    "交易編號": ["T001","T002","T003","T003","T004","T005","T005"],
    "商品": ["珍珠奶茶","珍珠奶茶","雞排","雞排","紅茶","蛋餅","蛋餅"],
    "金額": [65,65,90,90,40,55,55]
}

df = pd.DataFrame(data)

### 顯示原始資料
print("原始資料：")
df
```

## duplicated()：檢查整列是否重複

```python
df.duplicated()
# False 代表重複的第二項
```

```python
### subset：指定用哪個欄位判斷重複：真實資料不一定整列完全相同。
df.duplicated(subset=["交易編號"])
```

```python
### keep=False：把所有重複資料都找出來
# 如果想把重複交易的第一次和後續出現全部標記出來，可以使用
df.duplicated(
    subset=["交易編號"],
    keep=False
)
```

## drop_duplicates()：刪除重複資料


```python
# 如果我們認為「交易編號」應該唯一，可以指定：
df = df.drop_duplicates(
    subset=["交易編號"], # 如果交易編號重複，只保留第一次出現的資料。
    keep="first"
    # keep有三種
    # first：保留第一筆
    # last：保留最後一筆
    # False：重複的全部刪除
)
df
```

## 練習：會員消費紀錄重複資料

請完成以下任務：

1. 使用 `duplicated()` 檢查是否有**整列完全相同**的資料。
2. 只根據 `"交易編號"` 判斷是否重複，觀察哪些資料為 `True`。
3. 使用 `keep=False`，把 `"交易編號"` 有重複的資料**全部標記出來**。
4. 根據 `"交易編號"` 刪除重複資料，重複時**保留第一筆**。
5. 重新建立原始 `df`，這次根據 `"交易編號"` 刪除重複資料，改成**保留最後一筆**。
6. 比較第 4 題與第 5 題的結果，觀察 `keep="first"` 與 `keep="last"` 的差異。
