# 排序

## sort_values()：按照欄位內容排序

- by：按照哪個欄位排序
- ascending=True：由小到大
- ascending=False：由大到小

```python
import pandas as pd

## 建立學生成績資料
data = {
    "姓名": ["小明", "小美", "小華", "小安", "小傑"],
    "國文": [78, 92, 85, 67, 88],
    "數學": [90, 76, 88, 95, 82]
}

df = pd.DataFrame(data)
df
```

```python
## 按照數學成績由小到大排序
result = df.sort_values(
    by="數學",
    ascending=True
)

print("\n數學成績由小到大：")
result
```

```python

## 按照數學成績由大到小排序
result = df.sort_values(
    by="數學",
    ascending=False
)

print("\n數學成績由大到小：")
result
```

## 多欄位排序

實際資料分析很少只有一個排序規則。例如先按照「狀態」排序，如果狀態相同，再按照「實付金額」由高到低排序。

```python
import pandas as pd

## 建立訂單資料
data = {
    "訂單編號": ["A001", "A002", "A003", "A004", "A005", "A006"],
    "餐點金額": [180, 320, 250, 450, 150, 380],
    "外送費": [30, 0, 40, 0, 30, 20],
    "折扣": [20, 50, 0, 80, 10, 30],
    "狀態": ["已完成", "已完成", "已取消", "已完成", "已取消", "已完成"]
}

df = pd.DataFrame(data)

## 計算實付金額
df["實付金額"] = (df["餐點金額"] + df["外送費"] - df["折扣"] )

## 使用兩個欄位排序
result = df.sort_values(
    by=["狀態", "實付金額"],
    ascending=[True, False]
)
result
```

## nlargest()：找最大的前幾筆

```python
# 實付金額最高的 3 筆訂單是哪幾筆？
import pandas as pd

## 建立訂單資料
data = {
    "訂單編號": ["A001", "A002", "A003", "A004", "A005", "A006"],
    "餐點金額": [180, 320, 250, 450, 150, 380],
    "外送費": [30, 0, 40, 0, 30, 20],
    "折扣": [20, 50, 0, 80, 10, 30],
    "狀態": ["已完成", "已完成", "已取消", "已完成", "已取消", "已完成"]
}

df = pd.DataFrame(data)

## 計算實付金額
df["實付金額"] = (df["餐點金額"] + df["外送費"] - df["折扣"] )

## 找出實付金額最高的 3 筆
df.nlargest(3, "實付金額")
```

## nsmallest()：找最小的前幾筆

```python
# 實付金額最高的 3 筆訂單是哪幾筆？
import pandas as pd

## 建立訂單資料
data = {
    "訂單編號": ["A001", "A002", "A003", "A004", "A005", "A006"],
    "餐點金額": [180, 320, 250, 450, 150, 380],
    "外送費": [30, 0, 40, 0, 30, 20],
    "折扣": [20, 50, 0, 80, 10, 30],
    "狀態": ["已完成", "已完成", "已取消", "已完成", "已取消", "已完成"]
}

df = pd.DataFrame(data)

## 計算實付金額
df["實付金額"] = (df["餐點金額"] + df["外送費"] - df["折扣"] )

## 找出實付金額最高的 3 筆
df.nsmallest(3, "實付金額")
```
