### 使用原本的欄位計算新欄位

````python
import pandas as pd

### 建立訂單資料
data = {
    "訂單編號": ["A001", "A002", "A003", "A004"],
    "餐點金額": [180, 320, 250, 450],
    "外送費": [30, 0, 40, 0],
    "折扣": [20, 50, 0, 80],
    "狀態": ["完成", "完成", "取消", "完成"]
}

df = pd.DataFrame(data)
### 計算每筆訂單的實付金額
df["實付金額"] = (df["餐點金額"] + df["外送費"] - df["折扣"])
df
````

### 使用條件建立True/False欄位

````python
import pandas as pd

### 建立訂單資料
data = {
    "訂單編號": ["A001", "A002", "A003", "A004"],
    "餐點金額": [180, 320, 250, 450],
    "外送費": [30, 0, 40, 0],
    "折扣": [20, 50, 0, 80],
    "狀態": ["完成", "完成", "取消", "完成"]
}

df = pd.DataFrame(data)
### 計算每筆訂單的實付金額
df["實付金額"] = (df["餐點金額"] + df["外送費"] - df["折扣"])
### 判斷是否為大額訂單
df["大額訂單"] = df["實付金額"] >= 300

df[["訂單編號", "實付金額", "大額訂單"]]
````

### replace()：替換指定內容

資料清洗時很常遇到同一件事情有不同寫法。
例如：
```
完成
取消
處理中
```
但系統希望統一成：
```
已完成
已取消
處理中
```

````python

import pandas as pd

### 建立訂單資料
data = {
    "訂單編號": ["A001", "A002", "A003", "A004"],
    "餐點金額": [180, 320, 250, 450],
    "外送費": [30, 0, 40, 0],
    "折扣": [20, 50, 0, 80],
    "狀態": ["完成", "完成", "取消", "完成"]
}

df = pd.DataFrame(data)

### 替換指定內容
df["狀態"] = df["狀態"].replace({
    "完成": "已完成",
    "取消": "已取消"
})
df
````

### map()：按照對照表轉換資料

````python
import pandas as pd

### 建立訂單資料
data = {
    "訂單編號": ["A001", "A002", "A003", "A004"],
    "餐點金額": [180, 320, 250, 450],
    "外送費": [30, 0, 40, 0],
    "折扣": [20, 50, 0, 80],
    "狀態": ["完成", "完成", "取消", "完成"],
    "星期代碼": [0, 1, 2, 3]
}

df = pd.DataFrame(data)

### 建立星期對照表
weekday_map = {
    0: "星期一",
    1: "星期二",
    2: "星期三",
    3: "星期四",
    4: "星期五",
    5: "星期六",
    6: "星期日"
}


### 使用 map 將星期代碼轉換成中文
df["星期"] = df["星期代碼"].map(
    weekday_map
) # 根據 weekday_map 的對照規則，將 0、1、2、3 轉換成對應的星期文字
df
````
