小明正在準備升學，希望了解臺灣各大專校院的科系分布。請從政府資料開放平臺取得大專校院科系資料，使用 Pandas 整理資料，找出特定縣市、特定科系以及學生人數較多的科系，最後輸出分析結果。

- 資料集名稱：[大專校院各校科系別概況](https://data.gov.tw/dataset/9621)

### 下載資料

```python
import requests
import pandas as pd

### 設定 CSV 網址
# url = "這裡貼上政府資料開放平臺的 CSV 下載網址"
url = "https://stats.moe.gov.tw/files/opendata/sdata.csv"

headers = {
    "User-Agent": "Mozilla/5.0"
}


### 下載 CSV
response = requests.get(
    url,
    headers=headers,
    timeout=30
)

response.raise_for_status() # 如果請求失敗 (4xx、5xx)，直接拋出例外
print("狀態碼：", response.status_code)

### 儲存 CSV
with open("../pandas_datasets/大專院科系資料.csv", "wb") as file:
    file.write(response.content)

print("CSV 下載完成")
```

### 使用 Pandas 讀取 CSV

```python
df = pd.read_csv("../pandas_datasets/大專院科系資料.csv")

### 查看前 5 筆資料
df.head()
```

### 檢查缺失值與重複值

```python
print("\n缺失值：", df.isna().sum())
print("-"*40)
print(df.duplicated().sum())

### 移除完全重複的資料
df = df.drop_duplicates()

df
```

### 找出高雄市所有科系

```python
kaohsiung = df.loc[
    df["縣市名稱"] == "50 高雄市",
    [
        "學校名稱",
        "科系名稱",
        "學生數",
        "教師數"
    ]
]

print("\n高雄市大專校院科系：")
kaohsiung
```

### 找出名稱包含「資訊」的科系

```python
information = df.loc[
    df["科系名稱"].str.contains(
        "資訊",
        na=False
    ),
    [
        "學校名稱",
        "科系名稱",
        "學生數",
        "教師數",
        "縣市名稱"
    ]
]

print("\n名稱包含「資訊」的科系：")
information
```

### 找出學生人數大於等於 500 人的科系

```python
### 將學生數轉換成數字
df["學生數"] = pd.to_numeric(
    df["學生數"],
    errors="coerce" # 無法轉換的東西補上NaN
)

large_department = df.loc[
    df["學生數"] >= 500,
    [
        "學校名稱",
        "科系名稱",
        "學生數",
        "教師數",
        "縣市名稱"
    ]
]

print("\n學生人數大於等於 500 人的科系：")
large_department
```

### 新增「大型科系」欄位

```python
df["大型科系"] = "否"

df.loc[
    df["學生數"] >= 500,
    "大型科系"
] = "是"
```

### 保留需要的欄位&修改部分欄位名稱

```python
result = df[
    [
        "學校名稱",
        "科系名稱",
        "日間∕進修別",
        "等級別",
        "學生數",
        "教師數",
        "縣市名稱",
        "大型科系"
    ]
]


### 修改部分欄位名稱
result = result.rename(columns={
    "縣市名稱": "縣市",
    "日間∕進修別": "就讀類型"
})

result
```

### 輸出 CSV

```python
result.to_csv(
    "../pandas_output/大專校院科系整理.csv",
    index=False,
    encoding="utf-8-sig"
)

information.to_csv(
    "../pandas_output/資訊相關科系.csv",
    index=False,
    encoding="utf-8-sig"
)
```
