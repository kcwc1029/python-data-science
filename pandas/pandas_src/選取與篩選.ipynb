### 建立資料

```python
import pandas as pd

df = pd.DataFrame({
    "訂單編號": ["A001", "A002", "A003", "A004", "A005"],
    "餐點": ["牛肉麵", "雞腿飯", "火鍋", "水餃", "咖哩飯"],
    "餐點金額": [180, 120, 350, 90, 250],
    "狀態": ["完成", "完成", "未完成", "取消", "完成"],
    "付款方式": ["現金", "信用卡", "信用卡", "現金", "LINE Pay"]
})

df
```

### 選欄位：「我只想看哪些資料？」

```python
result_Series = df["餐點金額"] 
result_Series
```

```python
result_DataFrame = df[["訂單編號", "餐點金額"]]
result_DataFrame
```

### loc 與 iloc：「我要指定哪些位置的資料？」

- df.loc[我要哪些列, 我要哪些欄]`：看標籤、名稱、條件
    - `
- iloc：看第幾列、第幾欄


```python
result = df.loc[df["餐點金額"] >= 250, ["訂單編號", "餐點金額"]]
result
```

```python
result = df.iloc[:3, :2] # 我要前 3 筆資料，而且只看前 2 欄。
result
```

### 布林篩選：「哪些資料符合我的條件？」

```python
# 情境：餐廳想找出所有「已完成，而且餐點金額 >= 150」的訂單。
# 可以先拆成兩個條件：
# 1. df["狀態"] == "完成"
# 2. df["餐點金額"] >= 150
# 3. 再把兩個條件使用 & 結合

result = df[
    (df["狀態"] == "完成") &
    (df["餐點金額"] >= 150)
]

result
```
