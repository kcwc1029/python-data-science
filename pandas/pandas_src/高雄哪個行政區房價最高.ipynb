### 只留下這次會使用的欄位

原始資料欄位很多，但這個問題根本用不到全部。

````python
### 保留這次分析需要的欄位
df_new = df[
    [
        "行政區",
        "交易標的種類",
        "房地總價",
        "單價元平方公尺",
        "建物型態",
        "代表建號屋齡",
        "電梯"
    ]
]

df_new.head()
````

## 得知「哪個行政區平均單價最高？」

那就需要
```
行政區
單價 元平方公尺
```

因此缺少行政區或單價的資料無法參與這次分析

````python
### 缺少行政區或單價的資料無法參與這次分析
df_new = df.dropna(
    subset=[
        "行政區",
        "單價元平方公尺"
    ]
)
df_new.shape
````

````python
### 計算各行政區平均單價
district_price = (
    df_new
    .groupby("行政區")["單價元平方公尺"]
    .mean()
    .reset_index() # 將行政區從 index 變回一般欄位，同時將結果轉成 DataFrame
)


### 按照平均單價由高到低排序
district_price = district_price.sort_values(
    "單價元平方公尺",
    ascending=False
)

district_price.head()
````

## 把平方公尺換成「每坪」

政府資料提供的是「單價元平方公尺」，但台灣講房價通常會說坪數。

1 坪約等於 3.305785 平方公尺，因此：


````python
### 將每平方公尺單價換算成每坪單價
district_price["單價萬元每坪"] = (
    district_price["單價元平方公尺"]
    * 3.305785
    / 10000
)
district_price.head()
````

## 如果只想拿第一名：

````python
### 找出平均單價最高的行政區
highest_district = district_price.iloc[0]["行政區"]
highest_price = district_price.iloc[0]["單價萬元每坪"]

print("平均單價最高行政區：", highest_district)
print("平均單價：", round(highest_price, 2), "萬元/坪")
````
