### 下載資料

```python
import requests
import pandas as pd


### 設定 API 網址
url = "https://tdx.transportdata.tw/api/basic/v2/Bike/Station/City/Taipei?$format=JSON"

headers = {
    "User-Agent": "Mozilla/5.0"
}


### 發送 HTTP 請求
response = requests.get(url, headers=headers, timeout=30)
response.raise_for_status() # 如果請求失敗 (4xx、5xx)，直接拋出例外

print("狀態碼：", response.status_code)


### 將 JSON 轉換成 Python 資料
data = response.json()

print("\n資料筆數：")
print(len(data))


### 將 JSON 轉換成 DataFrame
df = pd.json_normalize(data)
df
```

### 找出地址包含「復興南路」的站點

```python
### 找出地址包含「復興南路」的站點
Fuxing = df.loc[
    df["StationAddress.Zh_tw"].str.contains(
        "復興南路",
        na=False # 如果遇到缺失值 NaN，就把它當成「沒有符合條件」(False)
    )
]

Fuxing
```

### 找出容量超過 90 台的站點

```python
large_station = df.loc[
    df["BikesCapacity"] >= 90,
]
large_station
```

### 新增「大型站點」欄位 (容量 ≥ 70 為「是」)。

```python
### 新增大型站點欄位
df["大型站點"] = "否"

df.loc[
    df["BikesCapacity"] >= 70,
    "大型站點"
] = "是"
```

### 將所有大型站點輸出成 大型YouBike站點.csv


```python
### 篩選大型站點
result = df.loc[
    df["大型站點"] == "是",
    [
        "BikesCapacity",
        "StationName.Zh_tw",
        "StationAddress.Zh_tw",
        "StationPosition.PositionLon",
        "StationPosition.PositionLat",
        "大型站點"
    ]
]


### 修改中文欄位名稱
result = result.rename(columns={
    "BikesCapacity": "可停放車位數",
    "StationName.Zh_tw": "站點名稱",
    "StationAddress.Zh_tw": "站點地址",
    "StationPosition.PositionLon": "經度",
    "StationPosition.PositionLat": "緯度",
    "大型站點": "大型站點"
})


### 顯示整理後資料
print("\n大型 YouBike 站點：")
result.head()
```

```python
### 輸出 CSV
result.to_csv(
    "../pandas_output/大型YouBike站點.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n已產生：大型YouBike站點.csv")
```
