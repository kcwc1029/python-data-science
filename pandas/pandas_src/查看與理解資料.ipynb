```python
df.head()                 # 前 5 列
df.tail()                 # 後 5 列
df.sample(5)              # 隨機抽樣，避免只看開頭
df.shape                  # 列數與欄數
df.columns                # 欄名
df.dtypes                 # 各欄型態
df.info()                 # 型態、非空值、記憶體摘要
df.describe()             # 數值統計
df.describe(include="all")
df.isna().sum()           # 每欄缺失量
df.nunique()              # 每欄不重複值數量
df["類別"].value_counts(dropna=False)
```
