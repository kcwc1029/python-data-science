### how：決定怎麼合併

merge() 有一個`how`參數，用來決定「如果兩張表的資料沒有完全對得上，要留下哪些資料？」

常見有四種：

| `how`     | 意義                 |
| --------- | -------------------- |
| `"inner"` | 只保留兩邊都有的資料 |
| `"left"`  | 左表全部保留         |
| `"right"` | 右表全部保留         |
| `"outer"` | 左右兩邊全部保留     |


```python
### 建立訂單資料
orders = pd.DataFrame({
    "店家編號": ["S01", "S02", "S03"],
    "餐點": ["雞排", "珍珠奶茶", "牛肉麵"]
})

### 建立店家資料
shops = pd.DataFrame({
    "店家編號": ["S01", "S02", "S04"],
    "店名": ["豪大雞排", "珍煮丹", "早餐店"]
})
```

```python
# inner：只留下兩邊都有的
result = orders.merge(
    shops,
    on="店家編號",
    how="inner"
)
result
```

```python
# left：左表全部留下
result = orders.merge(
    shops,
    on="店家編號",
    how="left"
)
result
```

```python
# outer：兩邊全部留下
result = orders.merge(
    shops,
    on="店家編號",
    how="outer"
)
result
```
