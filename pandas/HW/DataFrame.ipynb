### Problem. 建立一個 DataFrame

請使用 Pandas 建立一個名為 menu 的 DataFrame，內容如下：

| 品項 | 價格 | 庫存 |
| -- | -: | -: |
| 紅茶 | 30 | 20 |
| 綠茶 | 35 | 15 |
| 奶茶 | 50 | 10 |


- 建立 DataFrame。
- 印出 menu。
- 印出 menu.shape。

```python
import pandas as pd

menu = pd.DataFrame({
    "品項": ["紅茶", "綠茶", "奶茶"],
    "價格": [30, 35, 50],
    "庫存": [20, 15, 10]
})

print(menu)
print(menu.shape)
```

### Problem. 取出「價格」欄位

```python
import pandas as pd

menu = pd.DataFrame({
    "品項": ["紅茶", "綠茶", "奶茶"],
    "價格": [30, 35, 50],
    "庫存": [20, 15, 10]
})

# 請分別使用以下兩種方式取得「價格」欄位：
a = menu["價格"]
b = menu[["價格"]]

print(a.shape)
print(b.shape)
```
